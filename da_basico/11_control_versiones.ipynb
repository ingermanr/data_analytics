## Introducción al Control de Versiones

El control de versiones es esencial para cualquier proyecto de análisis de datos.

### Que es?
El control de versiones es una practica para registrar y gestionar cambios en archivos a lo largo del tiempo.

### Para que sirve?
Te permite trabajar con seguridad: puedes volver a versiones estables, entender que cambió y colaborar sin pisar trabajo de otras personas.

### Como se usa?
1. Inicializas o clonas un repositorio
2. Guardas avances en commits con mensajes claros
3. Sincronizas cambios con un remoto cuando trabajas en equipo

````python
print("=" * 70)
print("¿QUÉ ES CONTROL DE VERSIONES?".center(70))
print("=" * 70)

print("""
📚 CONTROL DE VERSIONES

Sistema que registra los cambios en archivos a lo largo del tiempo,
permitiendo recuperar versiones anteriores cuando sea necesario.

¿Por qué es importante?
  ✓ Historial completo de cambios
  ✓ Colaboración en equipo sin conflictos
  ✓ Recuperación de versiones anteriores
  ✓ Experimentación segura (branches)
  ✓ Backup automático
  ✓ Trazabilidad (quién, qué, cuándo, por qué)

Escenarios comunes:
  ❌ SIN control de versiones:
     • proyecto_final.py
     • proyecto_final_v2.py
     • proyecto_final_REALLY_FINAL.py
     • proyecto_final_REALLY_FINAL_ahora_si.py
     → ¿Cuál es la versión correcta?
     → ¿Qué cambió entre versiones?
  
  ✓ CON control de versiones:
     • proyecto.py (único archivo)
     • Historial completo en Git
     • Mensajes descriptivos de cada cambio
     • Posibilidad de volver a cualquier versión

Tipos de sistemas:
  1. LOCAL: En tu computadora (RCS)
  2. CENTRALIZADO: Servidor central (SVN, Perforce)
  3. DISTRIBUIDO: Copia completa en cada usuario (Git, Mercurial)

Git es el estándar de la industria:
  • Creado por Linus Torvalds (2005)
  • Sistema distribuido
  • Rápido y eficiente
  • Branching y merging poderosos
  • Usado por millones de desarrolladores
""")

print("\n" + "=" * 70)
print("CONCEPTOS CLAVE DE GIT".center(70))
print("=" * 70)

print("""
🔑 TERMINOLOGÍA GIT

REPOSITORY (Repositorio):
  Carpeta que contiene tu proyecto y todo su historial.
  Tiene una carpeta oculta .git/ con toda la información.

COMMIT:
  Snapshot (fotografía) del estado de tu proyecto en un momento.
  Cada commit tiene un ID único (hash SHA-1).
  Incluye mensaje descriptivo de los cambios.

BRANCH (Rama):
  Línea independiente de desarrollo.
  La rama principal se llama 'main' o 'master'.
  Útil para desarrollar features sin afectar el código principal.

MERGE:
  Combinar cambios de una rama en otra.
  Puede tener conflictos que requieren resolución manual.

REMOTE (Remoto):
  Versión del repositorio alojada en servidor (GitHub, GitLab).
  'origin' es el nombre por defecto del remoto principal.

CLONE:
  Copiar un repositorio remoto a tu máquina local.

PULL:
  Descargar cambios del repositorio remoto.

PUSH:
  Subir tus commits al repositorio remoto.

STAGING AREA (Área de preparación):
  Zona intermedia donde preparas los archivos antes del commit.
  'git add' mueve archivos al staging area.

Estados de archivos:
  1. UNTRACKED: Git no sigue este archivo
  2. MODIFIED: Archivo cambiado pero no en staging
  3. STAGED: Archivo listo para commit
  4. COMMITTED: Cambios guardados en el repositorio
""")
````

## Git Básico: Comandos Esenciales

### Que es?
Es el conjunto minimo de comandos Git para el flujo diario de trabajo.

### Para que sirve?
Dominar estos comandos evita errores comunes, mejora velocidad de trabajo y reduce friccion cuando colaboras.

### Como se usa?
Piensalo como ciclo: revisar estado -> preparar cambios -> confirmar cambios -> sincronizar con remoto.

````python
print("=" * 70)
print("GIT: COMANDOS ESENCIALES".center(70))
print("=" * 70)

print("""
🚀 FLUJO BÁSICO DE GIT

1. CONFIGURACIÓN INICIAL (una sola vez)
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

# Configurar tu identidad
git config --global user.name "Tu Nombre"
git config --global user.email "tu@email.com"

# Ver configuración
git config --list

# Configurar editor por defecto (opcional)
git config --global core.editor "code --wait"  # VSCode
git config --global core.editor "nano"          # Nano


2. INICIAR UN REPOSITORIO
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

# Crear nuevo repositorio en carpeta actual
git init

# Clonar repositorio existente
git clone https://github.com/usuario/proyecto.git
git clone https://github.com/usuario/proyecto.git mi-carpeta


3. WORKFLOW DIARIO
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

# Ver estado actual
git status                     # Estado detallado
git status -s                  # Estado resumido

# Agregar archivos al staging area
git add archivo.py             # Agregar archivo específico
git add .                      # Agregar todos los cambios
git add *.py                   # Agregar todos los .py
git add carpeta/               # Agregar toda la carpeta

# Quitar del staging (sin perder cambios)
git restore --staged archivo.py

# Hacer commit
git commit -m "Mensaje descriptivo del cambio"
git commit -am "Mensaje"       # Add + commit (solo archivos rastreados)

# Ver historial
git log                        # Historial completo
git log --oneline              # Historial resumido
git log --graph --oneline --all  # Con gráfico de branches
git log -n 5                   # Últimos 5 commits
git log --author="Nombre"      # Commits de un autor

# Ver diferencias
git diff                       # Cambios no staged
git diff --staged              # Cambios staged
git diff archivo.py            # Diferencias en archivo específico


4. DESHACER CAMBIOS
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

# Descartar cambios en archivo (PELIGROSO - no se puede deshacer)
git restore archivo.py
git checkout -- archivo.py     # Versión antigua del comando

# Modificar último commit (antes de push)
git commit --amend -m "Nuevo mensaje"

# Volver a commit anterior (mantiene cambios)
git reset HEAD~1

# Volver a commit anterior (descarta cambios - PELIGROSO)
git reset --hard HEAD~1

# Revertir un commit creando nuevo commit
git revert <commit-hash>


5. TRABAJO CON REMOTOS
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

# Ver remotos configurados
git remote -v

# Agregar remoto
git remote add origin https://github.com/usuario/repo.git

# Subir cambios (primera vez)
git push -u origin main        # -u establece tracking

# Subir cambios (siguientes veces)
git push

# Descargar cambios
git pull                       # fetch + merge
git fetch                      # Solo descarga, no fusiona


6. BRANCHES (RAMAS)
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

# Ver ramas
git branch                     # Ramas locales
git branch -a                  # Todas las ramas (local + remoto)

# Crear rama
git branch feature-nueva

# Cambiar de rama
git checkout feature-nueva
git switch feature-nueva       # Comando más nuevo

# Crear y cambiar en un solo comando
git checkout -b feature-nueva
git switch -c feature-nueva

# Fusionar rama
git checkout main              # Ir a rama destino
git merge feature-nueva        # Fusionar feature-nueva en main

# Eliminar rama
git branch -d feature-nueva    # Solo si ya está fusionada
git branch -D feature-nueva    # Forzar eliminación


7. CONSULTAS Y BÚSQUEDA
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

# Buscar en archivos
git grep "texto a buscar"

# Ver quién modificó cada línea
git blame archivo.py

# Ver contenido de commit específico
git show <commit-hash>

# Ver archivos en commit
git show <commit-hash> --name-only
""")
````

## .gitignore: Archivos a Ignorar

### Que es?
El archivo .gitignore define que archivos o carpetas Git debe ignorar.

### Para que sirve?
Evita subir archivos temporales, credenciales o datos pesados que no aportan al versionado del codigo.

### Como se usa?
1. Crear .gitignore en la raiz del proyecto
2. Agregar patrones por tipo de archivo/carpeta
3. Mantenerlo actualizado segun evolucione el proyecto

````python
print("=" * 70)
print(".gitignore: ARCHIVOS A IGNORAR".center(70))
print("=" * 70)

print("""
🚫 ¿QUÉ ES .gitignore?

Archivo especial que indica a Git qué archivos/carpetas NO debe rastrear.

¿Qué NO debes subir a Git?
  • Archivos grandes de datos (.csv, .xlsx grandes)
  • Credenciales y contraseñas (.env, config.ini)
  • Archivos temporales (__pycache__, .DS_Store)
  • Carpetas de entornos virtuales (venv/, env/)
  • Archivos de configuración de IDE (.vscode/, .idea/)
  • Archivos compilados (*.pyc, *.pyo)
  • Checkpoints de notebooks (.ipynb_checkpoints/)
  • Logs (*.log)
  • Archivos de sistema (.DS_Store, Thumbs.db)
""")

# Crear ejemplo de .gitignore
gitignore_content = """# Python
*.pyc
*.pyo
*.pyd
__pycache__/
*.so
*.egg
*.egg-info/
dist/
build/

# Entornos virtuales
venv/
env/
ENV/
virtualenv/
.venv/

# Jupyter Notebook
.ipynb_checkpoints/
*/.ipynb_checkpoints/*

# IDEs
.vscode/
.idea/
*.swp
*.swo
*~

# Datos
*.csv
*.xlsx
*.xls
*.db
*.sqlite
*.sqlite3
data/raw/*
data/processed/*
!data/raw/.gitkeep
!data/processed/.gitkeep

# Credenciales
.env
.env.local
config.ini
secrets.json
credentials.json

# Sistema operativo
.DS_Store
Thumbs.db
desktop.ini

# Logs
*.log
logs/

# Outputs
outputs/
results/
*.png
*.jpg
*.pdf

# Modelos ML (pueden ser grandes)
*.pkl
*.h5
*.model
models/*.pkl

# Excepciones (usar ! para no ignorar)
!README.md
!requirements.txt
"""

# Guardar .gitignore de ejemplo
with open('ejemplo_gitignore.txt', 'w', encoding='utf-8') as f:
    f.write(gitignore_content)

print("\n✓ Archivo de ejemplo creado: ejemplo_gitignore.txt")
print("\n📋 CONTENIDO DEL .gitignore RECOMENDADO:")
print("-" * 70)
print(gitignore_content)

print("""
💡 PATRONES EN .gitignore:

*.txt        → Ignora todos los .txt
file.txt     → Ignora file.txt específico
carpeta/     → Ignora toda la carpeta
**/temp      → Ignora carpetas temp en cualquier nivel
!importante.txt  → Excepción: NO ignorar este archivo
*.log        → Todos los archivos .log
logs/*.log   → Solo .log en carpeta logs/
logs/**/*.log    → .log en logs/ y subcarpetas

🔧 CÓMO USAR .gitignore:

1. Crear archivo .gitignore en la raíz del proyecto
2. Agregar patrones de archivos a ignorar
3. Commitear el .gitignore:
   git add .gitignore
   git commit -m "Agregar .gitignore"

⚠️ IMPORTANTE:
Si ya commiteaste un archivo antes de agregarlo a .gitignore,
Git seguirá rastreándolo. Para dejarlo de rastrear:

git rm --cached archivo.txt       # Un archivo
git rm --cached -r carpeta/       # Una carpeta
git commit -m "Dejar de rastrear archivo"

🌐 PLANTILLAS .gitignore:
GitHub ofrece plantillas para diferentes lenguajes:
https://github.com/github/gitignore
""")
````

## Jupyter Notebooks en Git

### Que es?
Es el conjunto de buenas practicas para versionar notebooks Jupyter sin generar ruido en diffs.

### Para que sirve?
Mejora trazabilidad del codigo real y disminuye conflictos de merge por outputs o metadata.

### Como se usa?
1. Limpiar outputs antes de commit
2. Usar herramientas como nbstripout
3. Priorizar estructura clara y documentacion en celdas markdown

````python
print("=" * 70)
print("JUPYTER NOTEBOOKS Y CONTROL DE VERSIONES".center(70))
print("=" * 70)

print("""
📓 DESAFÍOS CON NOTEBOOKS EN GIT

Problemas:
  1. Archivos .ipynb son JSON complejos
  2. Outputs de celdas causan diffs enormes
  3. Metadata cambia en cada ejecución
  4. Difícil ver qué realmente cambió en el código
  5. Conflictos de merge complicados

Ejemplo de .ipynb:
{
  "cells": [
    {
      "cell_type": "code",
      "execution_count": 5,
      "metadata": {...},
      "outputs": [{...}],  ← Esto cambia siempre
      "source": ["print('hola')"]
    }
  ],
  "metadata": {...}  ← Metadata cambia
}
""")

print("\n" + "=" * 70)
print("SOLUCIONES Y MEJORES PRÁCTICAS".center(70))
print("=" * 70)

print("""
✅ OPCIÓN 1: LIMPIAR OUTPUTS ANTES DE COMMIT
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

Manual:
  • En Jupyter: Kernel → Restart & Clear Output
  • Luego hacer commit

Desde línea de comandos:
  # Instalar nbconvert
  pip install nbconvert
  
  # Limpiar un notebook
  jupyter nbconvert --clear-output --inplace notebook.ipynb
  
  # Limpiar todos los notebooks
  jupyter nbconvert --clear-output --inplace **/*.ipynb


✅ OPCIÓN 2: NBSTRIPOUT (AUTOMATIZADO)
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

Herramienta que limpia automáticamente antes de cada commit.

Instalación:
  pip install nbstripout

Configuración en repositorio:
  # Instalar filtro Git
  nbstripout --install
  
  # Esto crea en .git/config:
  [filter "nbstripout"]
    clean  = nbstripout
    smudge = cat
  
  # Y crea/modifica .gitattributes:
  *.ipynb filter=nbstripout

Ahora cada commit limpiará automáticamente los outputs 🎉

Desinstalar:
  nbstripout --uninstall


✅ OPCIÓN 3: JUPYTEXT (SINCRONIZACIÓN .py)
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

Convierte notebooks a scripts .py legibles.

Instalación:
  pip install jupytext

Uso:
  # Convertir notebook a .py
  jupytext --to py notebook.ipynb
  
  # Sincronización automática (crear ambos)
  jupytext --set-formats ipynb,py notebook.ipynb
  
  # Ahora al guardar .ipynb, se actualiza .py automáticamente

Ventajas:
  • .py es fácil de revisar en git diff
  • Mejor para code review
  • Puedes commitear solo el .py


✅ OPCIÓN 4: REVIEWNB (GITHUB APP)
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

App para GitHub que muestra diffs visuales de notebooks.

Instalación:
  1. Ir a https://www.reviewnb.com/
  2. Conectar con tu cuenta de GitHub
  3. Instalar en repositorios deseados

Ventajas:
  • Diffs visuales de notebooks
  • Comentarios en celdas específicas
  • Ver outputs en PRs
  • Gratis para repos públicos
""")

print("\n" + "=" * 70)
print("MEJORES PRÁCTICAS CON NOTEBOOKS".center(70))
print("=" * 70)

print("""
📋 RECOMENDACIONES:

1. ESTRUCTURA DE PROYECTO
   notebooks/
   ├── 01_exploracion.ipynb
   ├── 02_limpieza.ipynb
   └── 03_modelado.ipynb
   
   src/
   ├── data_processing.py      ← Código reutilizable aquí
   ├── models.py
   └── visualization.py

2. SEPARAR CÓDIGO DE EXPLORACIÓN
   ✓ Notebooks: exploración, análisis, visualización
   ✓ Scripts .py: funciones reutilizables, clases
   ✓ Importar en notebooks: from src.data_processing import *

3. COMMITS
   ✓ Limpiar outputs antes de commit
   ✓ Commits frecuentes con mensajes claros
   ✓ Un notebook por feature/análisis

4. COLABORACIÓN
   ✓ Evitar trabajar varios en mismo notebook
   ✓ Dividir en notebooks separados por persona
   ✓ Usar nbstripout para evitar conflictos

5. DOCUMENTACIÓN
   ✓ Markdown cells explicando cada sección
   ✓ Títulos claros con #, ##, ###
   ✓ Comentarios en código complejo

6. .gitignore PARA NOTEBOOKS
   # Ignorar checkpoints
   .ipynb_checkpoints/
   */.ipynb_checkpoints/*
   
   # Opcional: ignorar notebooks en carpeta temporal
   notebooks/scratch/
   notebooks/temp/
""")

# Crear script de ejemplo para limpiar notebooks
clean_script = """#!/bin/bash
# Script para limpiar todos los notebooks antes de commit

echo "Limpiando outputs de notebooks..."

# Buscar todos los .ipynb y limpiar
find . -name "*.ipynb" -not -path "*/.ipynb_checkpoints/*" \\-exec jupyter nbconvert --clear-output --inplace {} \\;

echo "✓ Notebooks limpiados"
"""

with open('clean_notebooks.sh', 'w', encoding='utf-8') as f:
    f.write(clean_script)

print("\n✓ Script de limpieza creado: clean_notebooks.sh")
print("\nUso:")
print("  chmod +x clean_notebooks.sh")
print("  ./clean_notebooks.sh")
````
