## 1. Pensamiento Analítico

###  ¿Qué es el Pensamiento Analítico?

No es solo "analizar datos". Es un enfoque sistemático para:
- **Descomponer** problemas complejos en partes manejables
- **Identificar** patrones y relaciones
- **Evaluar** información objetivamente
- **Sintetizar** conclusiones lógicas

### Framework de Pensamiento Analítico

Cuando te enfrentes a cualquier problema, hazte estas 5 preguntas en orden:


````python
print("=" * 70)
print("FRAMEWORK DE PENSAMIENTO ANALÍTICO".center(70))
print("=" * 70)

print("""
Las 5 Preguntas Fundamentales:

┌─────────────────────────────────────────────────────────────────────┐
│ 1. ¿QUÉ QUIERO SABER/RESOLVER?                                     │
│    Definir el problema con precisión                               │
│    Ejemplo: "¿Por qué cayeron las ventas?" → muy amplio            │
│    Mejor: "¿Qué factores explican la caída del 15% en ventas       │
│             del producto X en Q2 vs Q1?"                            │
└─────────────────────────────────────────────────────────────────────┘

┌─────────────────────────────────────────────────────────────────────┐
│ 2. ¿QUÉ DATOS NECESITO?                                            │
│    Identificar fuentes de información relevantes                    │
│    • ¿Qué tablas/datasets?                                         │
│    • ¿Qué período temporal?                                        │
│    • ¿Qué granularidad?                                            │
│    • ¿Qué contexto adicional?                                      │
└─────────────────────────────────────────────────────────────────────┘

┌─────────────────────────────────────────────────────────────────────┐
│ 3. ¿QUÉ PODRÍA EXPLICAR ESTO?                                      │
│    Generar hipótesis ANTES de analizar                             │
│    • Causas internas (precio, calidad, marketing)                  │
│    • Causas externas (competencia, economía, estacionalidad)       │
│    • Causas técnicas (bugs, tracking issues)                       │
│    No te cases con ninguna hipótesis todavía                       │
└─────────────────────────────────────────────────────────────────────┘

┌─────────────────────────────────────────────────────────────────────┐
│ 4. ¿QUÉ DICE LA EVIDENCIA?                                         │
│    Analizar datos objetivamente                                     │
│    • Validar/refutar cada hipótesis con datos                      │
│    • Buscar contraejemplos                                         │
│    • Cuantificar el impacto                                        │
│    Dejar que los datos guíen, no tus prejuicios                    │
└─────────────────────────────────────────────────────────────────────┘

┌─────────────────────────────────────────────────────────────────────┐
│ 5. ¿QUÉ DEBO HACER CON ESTA INFORMACIÓN?                           │
│    Del insight a la acción                                          │
│    • ¿Qué decisión o acción se recomienda?                         │
│    • ¿Cuál es el próximo paso?                                     │
│    • ¿Qué más necesito investigar?                                 │
│    Un análisis sin acción es entretenimiento, no trabajo           │
└─────────────────────────────────────────────────────────────────────┘

""")
````

## 2. Resolución de Problemas

### 🎯 Framework de Resolución de Problemas

Pensamiento analítico te ayuda a ENTENDER. Resolución de problemas te ayuda a ACTUAR.


### Que es?
Es un enfoque metodico para pasar de un problema detectado a una solucion implementable.

### Para que sirve?
Evita soluciones superficiales y reduce reincidencia del mismo problema.

### Como se usa?
Sigue los pasos del framework en secuencia y valida causa raiz antes de ejecutar cambios.

````python
print("=" * 70)
print("FRAMEWORK DE RESOLUCIÓN DE PROBLEMAS".center(70))
print("=" * 70)

print("""
📋 LOS 6 PASOS PARA RESOLVER CUALQUIER PROBLEMA

┌────────────────────────────────────────────────────────────┐
│                                                            │
│  1. DEFINIR EL PROBLEMA                                   │
│     ¿Qué está mal? ¿Cuál es el gap entre actual y ideal? │
│                                                            │
│           ↓                                                │
│                                                            │
│  2. ANALIZAR EL PROBLEMA                                  │
│     ¿Cuál es la causa raíz? (5 Whys, Fishbone)           │
│                                                            │
│           ↓                                                │
│                                                            │
│  3. GENERAR ALTERNATIVAS                                  │
│     ¿Qué opciones tengo? (Brainstorm sin juzgar)         │
│                                                            │
│           ↓                                                │
│                                                            │
│  4. EVALUAR Y DECIDIR                                     │
│     ¿Cuál es la mejor opción? (Criterios: impacto,       │
│     esfuerzo, riesgo, tiempo)                             │
│                                                            │
│           ↓                                                │
│                                                            │
│  5. IMPLEMENTAR                                           │
│     Ejecutar la solución (con plan de acción claro)       │
│                                                            │
│           ↓                                                │
│                                                            │
│  6. MONITOREAR Y AJUSTAR                                  │
│     ¿Funcionó? ¿Qué aprendimos? ¿Ajustes necesarios?     │
│                                                            │
└────────────────────────────────────────────────────────────┘

⚠️ ERROR COMÚN: Saltar del paso 1 al paso 5
   ("Veo un problema → implemento primera solución que se me ocurre")
   
✅ MEJOR PRÁCTICA: Seguir todos los pasos metódicamente

""")
````

### 📝 Ejemplo: Problema de Datos Faltantes


````python
print("=" * 70)
print("EJEMPLO: RESOLVIENDO PROBLEMA DE DATOS FALTANTES".center(70))
print("=" * 70)

problema_ejemplo = """
🚨 SITUACIÓN:
Dashboard de ventas muestra datos vacíos para últimas 2 semanas.
Manager pregunta: "¿Qué pasó? ¿Cómo lo arreglamos?"

─────────────────────────────────────────────────────────────

PASO 1: DEFINIR EL PROBLEMA
─────────────────────────────────────────────────────────────
❌ Mal: "Los datos no están"
✅ Bien: "Sales dashboard no muestra transacciones desde 15-marzo.
         Última fecha con datos: 14-marzo. Missing: ~5,000 transacciones
         estimadas (basado en volumen histórico)"

Gap: Expected 5,000 transactions / Actual: 0 transactions

Impact: Manager no puede trackear ventas → decisiones en blind

─────────────────────────────────────────────────────────────

PASO 2: ANALIZAR EL PROBLEMA (5 Whys)
─────────────────────────────────────────────────────────────

Why #1: ¿Por qué dashboard no muestra datos?
→ Porque query no retorna filas para fechas >14-marzo

Why #2: ¿Por qué query no retorna filas?
→ Porque la tabla sales_daily no tiene datos recientes

Why #3: ¿Por qué la tabla no tiene datos recientes?
→ Porque ETL job no corrió exitosamente

Why #4: ¿Por qué ETL job falló?
→ Porque schema cambió en source system (nueva columna)

Why #5: ¿Por qué schema change rompió ETL?
→ Porque nuestro ETL asume schema fijo (no es resiliente)

✅ CAUSA RAÍZ IDENTIFICADA:
   ETL pipeline no es resiliente a schema changes

─────────────────────────────────────────────────────────────

PASO 3: GENERAR ALTERNATIVAS
─────────────────────────────────────────────────────────────

ALTERNATIVA A: Quick fix
  • Manualmente correr ETL con schema ajustado
  • Backfill datos faltantes
  • PRO: Rápido (2 horas)
  • CON: Se romperá de nuevo en próximo schema change

ALTERNATIVA B: Patch el ETL
  • Hardcodear manejo de la nueva columna
  • Backfill datos
  • PRO: Relativamente rápido (4 horas)
  • CON: Frágil, se romperá con futuros cambios

ALTERNATIVA C: Hacer ETL resiliente
  • Implementar schema detection automático
  • Manejo flexible de columnas nuevas/faltantes
  • Alertas cuando hay schema changes
  • Backfill datos
  • PRO: Solución robusta a largo plazo
  • CON: Más tiempo (2 días)

ALTERNATIVA D: Hybrid
  • Quick fix para tener datos HOY (alternativa A)
  • Trabajar en solución robusta paralelamente (alternativa C)
  • PRO: Balance de urgencia y solidez
  • CON: Doble esfuerzo

─────────────────────────────────────────────────────────────

PASO 4: EVALUAR Y DECIDIR
─────────────────────────────────────────────────────────────
"""

print(problema_ejemplo)

# Matriz de decisión
decision_matrix = pd.DataFrame({
    'Alternativa': ['A: Quick fix', 'B: Patch', 'C: Resiliente', 'D: Hybrid'],
    'Tiempo': ['2h', '4h', '2 días', '2h + 2 días'],
    'Resuelve HOY': ['✅', '✅', '❌', '✅'],
    'Previene futuro': ['❌', '⚠️', '✅', '✅'],
    'Esfuerzo': ['Bajo', 'Medio', 'Alto', 'Alto'],
    'Riesgo': ['Alto', 'Medio', 'Bajo', 'Bajo'],
    'Score Total': ['5/10', '6/10', '8/10', '9/10']
})

print("\n📊 MATRIZ DE DECISIÓN:\n")
print(decision_matrix.to_string(index=False))

print("""

🎯 DECISIÓN: Alternativa D (Hybrid)

JUSTIFICACIÓN:
  • Urgente: Manager necesita datos HOY → Quick fix
  • Importante: Evitar problema recurrente → Solución robusta
  • Balance de corto y largo plazo

─────────────────────────────────────────────────────────────

PASO 5: IMPLEMENTAR
─────────────────────────────────────────────────────────────

PLAN DE ACCIÓN:

HOY (Día 1):
  09:00-09:30  Comunicar situación y plan a manager
  09:30-11:00  Quick fix: Ajustar ETL manualmente
  11:00-12:00  Backfill datos faltantes
  12:00-12:30  Validar que dashboard funciona
  13:00        Update a manager: "Datos restaurados"
  14:00-17:00  Empezar diseño de solución robusta

DÍA 2-3:
  • Implementar schema detection
  • Testing exhaustivo
  • Deploy a staging
  • Monitoring

DÍA 4:
  • Deploy a production
  • Documentar solución
  • Post-mortem con equipo

COMUNICACIÓN PROACTIVA:
  "Identificado issue de schema change en source system.
   
   Acción inmediata: Backfilling datos, dashboard restaurado a las 12pm.
   
   Acción preventiva: Implementando ETL resiliente para evitar
   recurrencia. ETA: 4 días.
   
   Root cause: Necesitamos mejor coordinación con source system team
   en schema changes. Proponiendo process improvement."

─────────────────────────────────────────────────────────────

PASO 6: MONITOREAR Y AJUSTAR
─────────────────────────────────────────────────────────────

MÉTRICAS DE ÉXITO:
  □ Dashboard con datos completos (VERIFICADO)
  □ ETL corriendo sin fallos >1 semana (PENDIENTE)
  □ Alertas funcionando correctamente (PENDIENTE)
  □ Zero schema-change incidents en próximo mes (PENDIENTE)

RETROSPECTIVA (después de 1 semana):
  ✅ ¿Qué fue bien?
     • Respuesta rápida restauró datos
     • Comunicación proactiva mantuvo confianza
     • Solución robusta previene recurrencia

  ⚠️ ¿Qué mejorar?
     • Debimos tener alertas antes (incident detection tardío)
     • Coordinación con source team es débil
     • Falta documentation de nuestro ETL

  🎯 ¿Qué haremos diferente?
     • Implementar data quality monitoring continuous
     • Weekly sync con source system team
     • Documentar todos los pipelines

APRENDIZAJE:
  "Invertir en resilience upfront habría evitado este incident.
   Technical debt has cost."

━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

💡 OBSERVACIONES CLAVE:

1. No nos quedamos en quick fix (aunque fue tentador)
2. Balanceamos urgencia vs solución sostenible
3. Comunicamos proactivo y transparente
4. Documentamos aprendizajes
5. Prevenimos futuras ocurrencias

ESTO ES RESOLUCIÓN DE PROBLEMAS PROFESIONAL.
""")
````

### 🛠️ Herramientas de Análisis de Causa Raíz

**5 Whys** (como vimos arriba)
- Simple y efectivo
- Mejor para problemas lineales

**Fishbone Diagram (Ishikawa)**
- Visual
- Categorías: People, Process, Technology, Data, Environment
- Mejor para problemas complejos con múltiples causas potenciales

**Pareto Analysis (80/20)**
- 80% de los problemas vienen de 20% de las causas
- Enfoca en el vital few

### 🎯 Principios de Comunicación Efectiva


### Que es?
Comunicar resultados es traducir analisis tecnico en mensajes claros para decision makers.

### Para que sirve?
Sin comunicacion efectiva, incluso un analisis correcto puede no generar accion ni impacto.

### Como se usa?

````python
print("=" * 70)
print("COMUNICACIÓN EFECTIVA DE RESULTADOS".center(70))
print("=" * 70)

print("""
🎯 REGLA DE ORO:

  "Conoce a tu audiencia"

No comunicas igual a:
  • CEO (quiere decisión + impacto en negocio)
  • Data Scientist (quiere metodología + detalles técnicos)
  • Product Manager (quiere insights accionables)
  • Ingeniero (quiere especificaciones claras)

─────────────────────────────────────────────────────────────

📊 ESTRUCTURA PIRAMIDAL (Pyramid Principle)

Empieza por la conclusión, luego justifica:

    ┌───────────────────────────┐
    │   CONCLUSIÓN/             │  ← Empieza aquí
    │   RECOMENDACIÓN           │
    └───────────────────────────┘
            ▲
            │
    ┌───────┴───────┬───────────┐
    │ KEY INSIGHT 1 │ KEY INS.2 │ KEY INSIGHT 3
    └───────────────┴───────────┴───────────────┘
            ▲
            │
    ┌───────┴───────┬───────────┐
    │ Evidencia 1A  │ Evid. 1B  │ Evidencia 1C ...
    └───────────────┴───────────┴───────────────┘

❌ MAL (Bottom-up):
"Primero analicé la tabla X, luego crucé con Y, después calculé Z...
 (10 minutos después) ...y por eso recomiendo W"

✅ BIEN (Top-down):
"Recomiendo W porque A, B, C. Déjame explicar cada punto..."

Si te interrumpen, ya dijiste lo importante.

─────────────────────────────────────────────────────────────

🎤 TRES VERSIONES DE CADA MENSAJE

Prepara 3 versiones:

1. ELEVATOR PITCH (30 segundos)
   "¿Qué encontraste?"
   → "El churn aumentó 3pp por cambio de precio. Recomiendo
      crear tier intermedio. Esperamos recuperar 2pp."

2. EXECUTIVE SUMMARY (2-3 minutos)
   "Cuéntame más"
   → Insight principal + 2-3 insights de soporte + recomendación
      + impacto estimado

3. DEEP DIVE (30+ minutos)
   "Quiero entender el análisis completo"
   → Metodología + todos los insights + análisis de sensibilidad
      + anexos técnicos

Empieza siempre con versión 1. Si hay interés, avanza a 2, luego 3.

NO fuerces versión 3 si audiencia solo quiere versión 1.

─────────────────────────────────────────────────────────────

📝 FRAMEWORK: SITUACIÓN → COMPLICACIÓN → RESOLUCIÓN

ESTRUCTURA NARRATIVA efectiva:

SITUACIÓN:
  "Nuestro objetivo es reducir churn a <5%"

COMPLICACIÓN:
  "Pero churn ha aumentado a 8% en Q1, el más alto en 2 años"

RESOLUCIÓN:
  "Identifiqué 3 causas. Primera acción: ajustar pricing.
   Esperamos retornar a 5% en Q2."

Esta estructura crea tensión y resolución → memorable

─────────────────────────────────────────────────────────────

💡 LENGUAJE EFECTIVO

❌ LENGUAJE TÉCNICO INNECESARIO:
"El coeficiente de correlación de Pearson entre revenue per user
 y session frequency es 0.78 con p-value <0.001"

✅ LENGUAJE ORIENTADO A NEGOCIO:
"Usuarios más activos gastan significativamente más (78% correlación).
 Incrementar engagement 10% → +$50K revenue mensual estimado"

❌ PASIVO Y VAGO:
"Se observó que las ventas disminuyeron"

✅ ACTIVO Y ESPECÍFICO:
"Las ventas cayeron 15% ($200K) en marzo"

❌ SOLO DESCRIPCIÓN:
"El 60% de usuarios no completa onboarding"

✅ DESCRIPCIÓN + IMPLICACIÓN:
"El 60% abandona onboarding → perdemos $300K en revenue potencial
 anual. Optimizar onboarding es top priority."

─────────────────────────────────────────────────────────────

🎨 VISUALIZACIÓN EN COMUNICACIÓN

Reglas de oro:

1. UN MENSAJE POR GRÁFICO
   Cada gráfico debe responder UNA pregunta específica

2. MENOS ES MÁS
   Elimina todo lo que no añade información

3. DESTACA LO IMPORTANTE
   Color, tamaño, anotaciones para guiar la vista

4. CONTEXTO SIEMPRE
   Comparación con baseline, benchmark, target

Ejemplos próximos...

""")
````

### Como se usa?
Organiza la presentacion en: contexto, hallazgos clave, impacto y siguiente paso recomendado.
